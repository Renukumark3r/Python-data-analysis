**Assignments 2**

```python
# Create a list and perform basic operations like append(), remove(), pop(), and slicing.
frutes=['apple','banana','orange']
print('list of frutes', frutes)
#append
frutes.append('pineapple')
print('Add one more frutes in the list',frutes)
#slicing
print('slicing the list',frutes[0:3])
#remove
frutes.remove('banana')
print('Remove the banana in the list',frutes)
#pop
frutes.pop(1)
print('After pop index(1)',frutes)
```

```python
#Iterate over a list using loops.
frutes_1=['apple','banana','orange']
for i in frutes_1:
  print(i)

```

**Assignment 4**

```python
#Dictionary Operations
student={'name':'Renu','age':25,'mark':88}
print(student)
#Adding
student['location']='chennai'
print('After add the location',student)
#Updating
student['age']=30
print('After updating the age',student)
# deleting
del student['mark']
print('After deleting the mark',student)
#Iterate through a dictionary using loops.
for key,value in student.items():
  print(key,value)
```
