**Assignment 5**

```python
#Set Operations
set_1={1,2,3,4}
set_2={5,6,7,2,3}
#Perform union
print("Print the all the value in the two set",set_1|set_2)
# Perform intersection
print("Print the same value in the two set",set_1 & set_2)
# Perform difference
print("Print the different value in the first set",set_1-set_2)
print("Print the different value in the second set",set_2-set_1)
#Understand the uniqueness property of sets
print("Print the uniqueness value ",set_1^set_2)
```
